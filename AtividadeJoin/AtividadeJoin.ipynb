# Aula — Join das Tabelas 5 (2012 e 2026) com a Tabela 1.1.1 do IBGE (2022)

Nesta atividade vamos:

1. Carregar as duas **Tabelas 5**, referentes à composição da desocupação por sexo em **2012** e **2026**.
2. Juntar as duas tabelas usando `Sigla`, `Código` e `Estado`.
3. Preparar a **Tabela 1.1.1 do IBGE (2022)**, usando a aba `2022` do arquivo Excel.
4. Cruzar as três tabelas pelo nome da Unidade da Federação (`Estado`).
5. Calcular algumas diferenças para facilitar a comparação entre os anos.

> Os arquivos utilizados estão no mesmo ZIP desta atividade.

````python
import pandas as pd
````

````python
# Arquivos da atividade
arquivo_2012 = r"H:\python\atividades\GitNovo\AtividadeJoin\Tabela5-sem_emprego_2012.csv"
arquivo_2026 = r"H:\python\atividades\GitNovo\AtividadeJoin\Tabela5-sem_emprego_2026.csv"
arquivo_ibge = r"H:\python\atividades\GitNovo\AtividadeJoin\Tabela 1.1.1.xlsx"

# Leitura das Tabelas 5
t2012 = pd.read_csv(arquivo_2012, sep=";", decimal=",")
t2026 = pd.read_csv(arquivo_2026, sep=";", decimal=",")

print("Tabela 2012:")
display(t2012.head())

print("Tabela 2026:")
display(t2026.head())
````

## 1. Juntando as duas Tabelas 5

A junção segue o modelo solicitado na aula:

```python
t2012.merge(t2026, on=["Sigla", "Código", "Estado"], how="inner")
```

Assim, cada Unidade da Federação fica em uma única linha, com os valores de 2012 e 2026.

````python
# Join das Tabelas 5
comp = t2012.merge(
    t2026,
    on=["Sigla", "Código", "Estado"],
    how="inner"
)

# Reorganizando as colunas
comp = comp[
    [
        "Sigla", "Código", "Estado",
        "Desocupados - homens (2012 T1)",
        "Desocupados - mulheres (2012 T1)",
        "Desocupados - homens (2026 T1)",
        "Desocupados - mulheres (2026 T1)"
    ]
]

print(f"Quantidade de UFs após o join: {len(comp)}")
display(comp)
````

````python
# Diferenças entre 2026 e 2012
comp["Variação homens (p.p.)"] = (
    comp["Desocupados - homens (2026 T1)"]
    - comp["Desocupados - homens (2012 T1)"]
)

comp["Variação mulheres (p.p.)"] = (
    comp["Desocupados - mulheres (2026 T1)"]
    - comp["Desocupados - mulheres (2012 T1)"]
)

display(comp)
````

## 2. Preparando a Tabela 1.1.1 do IBGE

A planilha Excel possui várias abas. Para esta atividade usamos a aba **`2022`**, que contém o número médio de horas semanais dedicadas aos cuidados de pessoas e/ou afazeres domésticos.

Como a tabela tem cabeçalhos em várias linhas, vamos localizar os dados a partir da linha em que aparecem as Unidades da Federação. Serão mantidas:

- `Estado`
- `Horas - total`
- `Horas - homem branco`
- `Horas - homem preto ou pardo`
- `Horas - mulher branca`
- `Horas - mulher preta ou parda`

A coluna `Horas - total` representa o número médio de horas semanais da população de 14 anos ou mais.

````python
# Leitura da aba 2022 sem cabeçalho para podermos tratar os títulos da tabela
ibge_raw = pd.read_excel(arquivo_ibge, sheet_name="2022", header=None)

# Os dados começam na linha 8 (índice 8)
ibge = ibge_raw.iloc[8:, [0, 1, 4, 5, 6, 7]].copy()

ibge.columns = [
    "Estado",
    "Horas - total",
    "Horas - homem branco",
    "Horas - homem preto ou pardo",
    "Horas - mulher branca",
    "Horas - mulher preta ou parda"
]

# Retira linhas que não são UFs
ufs = set(comp["Estado"])
ibge = ibge[ibge["Estado"].isin(ufs)].copy()

# Converte os valores para número
colunas_horas = [c for c in ibge.columns if c != "Estado"]
for coluna in colunas_horas:
    ibge[coluna] = pd.to_numeric(ibge[coluna], errors="coerce")

ibge = ibge.reset_index(drop=True)

print(f"Quantidade de UFs na Tabela 1.1.1: {len(ibge)}")
display(ibge)
````

## 3. Cruzando as três tabelas

Agora usamos o resultado das duas Tabelas 5 (`comp`) e fazemos um novo `merge` com a Tabela 1.1.1 do IBGE.

A chave de cruzamento é `Estado`, pois a Tabela 1.1.1 não possui as colunas `Sigla` e `Código` presentes nas Tabelas 5.

````python
# Cruzamento das três tabelas
cruzado = comp.merge(
    ibge,
    on="Estado",
    how="inner"
)

# Colunas em uma ordem mais fácil de ler
cruzado = cruzado[
    [
        "Sigla", "Código", "Estado",
        "Desocupados - homens (2012 T1)",
        "Desocupados - mulheres (2012 T1)",
        "Desocupados - homens (2026 T1)",
        "Desocupados - mulheres (2026 T1)",
        "Variação homens (p.p.)",
        "Variação mulheres (p.p.)",
        "Horas - total",
        "Horas - homem branco",
        "Horas - homem preto ou pardo",
        "Horas - mulher branca",
        "Horas - mulher preta ou parda"
    ]
]

print(f"Quantidade de UFs no cruzamento final: {len(cruzado)}")
display(cruzado)
````

## 4. Comparação ano a ano

Abaixo podemos observar as mudanças na composição da desocupação por sexo entre 2012 e 2026.

A variação é calculada em **pontos percentuais (p.p.)**:

`valor de 2026 - valor de 2012`.

````python
comparacao = cruzado[
    [
        "Sigla", "Estado",
        "Desocupados - homens (2012 T1)",
        "Desocupados - homens (2026 T1)",
        "Variação homens (p.p.)",
        "Desocupados - mulheres (2012 T1)",
        "Desocupados - mulheres (2026 T1)",
        "Variação mulheres (p.p.)"
    ]
].copy()

display(comparacao.sort_values("Variação mulheres (p.p.)", ascending=False))
````

## 5. Cruzamento com as horas de cuidados e afazeres domésticos

Para facilitar a leitura do resultado final, selecionamos as variáveis de desocupação e as horas semanais do IBGE.

````python
resultado_final = cruzado[
    [
        "Sigla", "Estado",
        "Desocupados - homens (2012 T1)",
        "Desocupados - mulheres (2012 T1)",
        "Desocupados - homens (2026 T1)",
        "Desocupados - mulheres (2026 T1)",
        "Variação homens (p.p.)",
        "Variação mulheres (p.p.)",
        "Horas - total",
        "Horas - homem branco",
        "Horas - homem preto ou pardo",
        "Horas - mulher branca",
        "Horas - mulher preta ou parda"
    ]
].copy()

display(resultado_final)
````

## 6. Exportando o resultado

Por fim, o resultado do cruzamento das três tabelas é salvo em CSV, usando `;` como separador, conforme o exemplo da aula.

````python
# Salvar o resultado final
arquivo_saida = "geral.csv"
resultado_final.to_csv(arquivo_saida, sep=";", decimal=",", index=False, encoding="utf-8-sig")

print(f"Arquivo salvo: {arquivo_saida}")
````
